# Lists

Lists are one of the most used data structures in python (and pretty much every programming language).
We can start with basic lists.

## Basic lists

Lists are denoted using the `[]` characters. Elements in the list are separated by `,`:

```python
[3,4,6,2,1,4,5]
```

We can certainly have multidimensional lists. We can create a "list of lists":

```python
[[2,3],[5,6],[4,3]]
```

We get the size (length) of the list with the `len` function:

```python
list=[3,4.5,6,2,2.1]
len(list)
```

We access individual elements by position using `[]`:

```python
list[3]
```

The negative index -1 gives the last item. Negative indexes go backwards.

```python
list[-1]
```

We can get the `type` of individual list elements:

```python
type(list[1])
```

The ranges can also be used to get subsets of elements in a list, for example from position 2 to 4:

```python
list[2:4]
```

Or, from potision 1 until the end:

```python
list[1:]
```

Or from the beginning until position 3:

```python
list[:3]
```

Lists and multidimensional lists can contain different types of data

```python
biglist=[['a','b'],[2,3],[True,False]]
biglist
```

```python
biglist[1]
```

And we can access elements inside the list of a list, for example the element in position 1 from the list on position 1:

```python
biglist[1][1]
```

## List operations

We can delete an item using the `del` operator:

```python
del list[2]
print(list)
```

We can append an item to the list using the `append` function:

```python
list.append(44)
print(list)
```

The `reverse` function will invert the order of the list:

```python
list.reverse()
print(list)
```

The `sort` function will order the elements in ascending order:

```python
list.sort()
print(list)
```

```python
list.sort(reverse=True)
print(list)
```

Modifications of liste elements can be done directly by assigning a specific cell `[]`.
The `count` function counts the number of elements having a given value.

```python
list[0]=3
list.count(3)
```

We can get the index of a specific value. it gets the first occurence.

```python
list.index(4.5)
```

Replace a value in the list

```python
list[0]=55
print(list)
```

We can replace several cells at once, by specifying a range, for example cells from position `2` until `4`:

```python
list[2:4]=[77,88]
print(list)
```

We can also copy an entire list by selecting the full range from the beginning to the end.

```python
copylist=list[:]
copylist
```

The `range` function generates a list, for example from 2 until 8:

```python
numbers=range(2,8)
numbers[-1]
```

```python
numbers[0]
```
