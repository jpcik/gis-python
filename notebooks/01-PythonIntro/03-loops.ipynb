# Loops

Iterating over collections and over ranges is an essential programming aspect.

We can strat by treating strings a collections of chars.

## Strings

Consider a string that is simply a list of strings. So essentially all our lists functions work on them:

```python
name='samantha'
len(name)
```

We can also access individual chars, etc: 

```python
print(name[2])
```

Now that we see that strings are lists, let's use a `for`loop to iterate over every chars:

```python
for ch in name:
    print(ch)
print(ch)
```

A classical loop pattern is to have a counter that is incremented on each iteration:

```python
count=0
for ch in name:
    print(ch,count)
    count=count+1
print(count)
```

As we saw before, we can generate lists with ranges:

```python
range(4)
```

Then we can iterate over a range, for example from `0` until `4`:

```python
for i in range(4):
    print(i)
```

Ranges can have more parameters, for example with a step = 3, and up to 14

```python
for i in range(2,14,3):
    print(i)
```

Another typical pattern is to iterate over the size of a collection:

```python
name='Aladdin'
for i in range(len(name)):
    print(name[i])
```

As in other programming languages, we can also do `while` loops that work in the usual way with a condition:

```python
c=1
while (c<10) :
    print(c)
    c=c+2
    
```

## List comprehensions

There are other ways to create lists. List comprehensions means to construct a list by saying *"what the list should contain"*.

Let's start with the range that we already know. This example is a range from `1` until `5`:

```python
range(1,5)
```

We can create an actual list based on a range:

```python
list(range(1,5))
```

Now we can create our first **list comprehension**. List comprehensions are denoted with the typical list brackets `[]`.

This expression means: *the list contains all elements `x` such that `x` is in a range from `1` until `5`*

```python
[x for x in range(1,5)]
```

Then we can make it more interesting. This example build a list but instead of putting `x`, we will put `x` mutiplied by 2:

```python
[x*2 for x in range(1,5)]
```

Another examle, this list consists of all `x` square, only for pair numbers form `1` until `5`:

```python
[x**2 for x in range(1,5) if x%2==0 ]
```

Lists can have more complex elements, for example in this list we create a tuple `(x,x**2)` for ech element in the range that mathces the given condition:

```python
[(x,x**2) for x in range(1,5) if x%2 == 0]
```

We can also take a list as a basis, and then create a comprehension based on that list.

```python
list=[3,2,6,5,8]
[(x,x**2) for x in list if x%2 == 0]
```

We can creathe comprehensions with any data types. For example with strings we can also create this list that converts all strings to uppercase, only if they start with `'V'`:

```python
cantons=['valais','vaud','jura']
[s.upper() for s in cantons if s.startswith('v')]
```
