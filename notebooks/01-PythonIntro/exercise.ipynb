How many countries?

```python
data = [line.strip().split(',') for line in lines]
```

```python
data[:5]
```

```python
data = data[1:]
```

```python
countries= []
for entry in data:
    if not entry[1] in countries :
      countries.append(entry[1])
        
countries.sort()
print(countries)
```

```python
len(countries)
```

```python
countries = set([entry[1] for entry in data])
```

```python
len(countries)
```
