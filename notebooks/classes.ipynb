# Classes

```python
class Vehicle(object):
    wheels=0
    
    def __init__(self,wheels,maxSpeed=0):
        self.wheels=wheels
        self.maxSpeed=maxSpeed
        
    def fasterThan(self,otherVehicle):
        return self.maxSpeed > otherVehicle.maxSpeed
  
```

```python
v1=Vehicle(2)
print(v1.maxSpeed)
```

```python
class Bike(Vehicle):
  def __init__(self,maxSpeed=0):
    self.wheels=2
    self.maxSpeed=maxSpeed


b1=Bike(30)
b1.wheels
b1.maxSpeed
```

```python
b2=Vehicle(4,80)

b1.fasterThan(b2)
```
