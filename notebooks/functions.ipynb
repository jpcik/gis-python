# Functions

```python
def calculateArea(length,height):
    return length*height

calculateArea(5,4)
```

```python
calculateArea('a',4)
```

```python
def countLetters(strings,letter):
    count=0
    for str in strings:
        count=count+str.count(letter)
    return count
```

```python
words=['day','pasta','lasagna']
countLetters(words,'a')
```

```python
sum(sum(c=='a' for c in w) for w in words)
```

```python
def printPerson(name,age=20,sex='Female'):
    print(f'Name is {name}, age: {age} and is {sex}')
```

```python
printPerson('Rita')
```

```python
printPerson('Lara',30)
```

```python
printPerson('Granit',32,'Male')
```

```python
printPerson(age=40,name='Carla')
```

```python
def printStudent(name,*hobbies,**grades):
    print(f'his name is: {name}')
    print('\nHis hobbies:') 
    for h in hobbies:
        print(h)
    print('\nhis grades:')
    for g in grades:
        print(g,grades[g])
```

```python
printStudent('Lucas','running','curling',math=5,chem=3)
```
