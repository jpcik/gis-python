# Functions

```python
def calculateArea(length,height):
    return length*height

calculateArea(5,4)
```

```python
calculateArea('a',4)
```

```python
def countLetters(strings,letter):
    count=0
    for str in strings:
        count=count+str.count(letter)
    return count
```

```python
countLetters(['day','pasta','lasagna'],'a')
```
